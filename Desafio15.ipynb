```julia
using Dates
using Base.Threads

agora_utc = Dates.now()
agora_brasilia = agora_utc - Hour(3)   # UTC → Brasília (UTC−3)
println("Horário de Brasília: ", agora_brasilia)
println("Threads disponíveis: ", nthreads())

# Função pesada de exemplo
function trabalho_pesado(x)
    s = 0.0
    for i in 1:10_000_00
        s += sin(i + x)
    end
    return s
end

# Execução em série
function serie()
    resultados = zeros(Float64, 8)
    for x in 1:8
        resultados[x] = trabalho_pesado(x)
    end
    return resultados
end

# Execução em paralelo usando threads
function paralelo()
    resultados = zeros(Float64, 8)
    @threads for x in 1:8
        resultados[x] = trabalho_pesado(x)
    end
    return resultados
end

println("\n--- Série ---")
@time r1 = serie()

println("\n--- Paralelo ---")
@time r2 = paralelo()

println("\nResultado série:")
println(r1)

println("\nResultado paralelo:")
println(r2)
```

O código usa múltiplas threads para paralelizar o processamento da função trabalho_pesado.
Primeiro, verificamos quantas threads estão disponíveis no computador usando nthreads().
Depois, criamos duas versões do loop: uma em série e outra em paralelo.

Na versão em série, cada valor de x entre 1 e 8 é processado um após o outro, o que torna
a execução mais lenta. Já na versão paralela, o comando @threads divide as iterações do
loop entre as threads disponíveis, permitindo que vários valores sejam processados ao
mesmo tempo.

Isso torna o tempo de execução significativamente menor, já que as threads compartilham
a mesma memória e podem executar cálculos simultaneamente. Assim como no exemplo anterior,
o paralelismo traz uma boa melhora de desempenho mesmo com poucas iterações.

Note que foi criada uma variável para printar a data e hora que este código foi executado.
